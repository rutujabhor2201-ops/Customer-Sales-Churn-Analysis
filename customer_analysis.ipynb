# Customer Sales & Churn Analysis

This notebook analyzes sales performance and customer churn using Python, Pandas, and Matplotlib.

**Objectives:**
- Analyze overall and product/region-wise sales performance.
- Identify sales trends and high-performing products/regions.
- Analyze customer churn and factors associated with churn.
- Generate visualizations and business recommendations.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load datasets (keep these CSV files in the same folder as this notebook)
sales_data = pd.read_csv('sales_data.csv')
customer_data = pd.read_csv('customer_data.csv')

print('Sales data shape:', sales_data.shape)
print('Customer data shape:', customer_data.shape)
```

## 1. Sales Data Overview

```python
print('Sales Data - First 5 Rows')
display(sales_data.head())
print('\nMissing values:')
display(sales_data.isnull().sum())
print('\nDuplicate rows:', sales_data.duplicated().sum())
sales_data['Date'] = pd.to_datetime(sales_data['Date'])
```

## 2. Overall Sales Performance

```python
total_sales = sales_data['Total_Sales'].sum()
average_sales = sales_data['Total_Sales'].mean()
total_quantity = sales_data['Quantity'].sum()
total_transactions = len(sales_data)

print('Total Transactions:', total_transactions)
print('Total Quantity Sold:', total_quantity)
print('Total Sales:', total_sales)
print('Average Sales per Transaction:', round(average_sales, 2))
```

## 3. Product-wise Sales Analysis

```python
product_sales = sales_data.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
product_quantity = sales_data.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
display(product_sales.to_frame('Total Sales'))
display(product_quantity.to_frame('Quantity Sold'))

print('Top Product:', product_sales.idxmax())
print('Top Product Sales:', product_sales.max())

plt.figure(figsize=(9,6))
product_sales.plot(kind='bar')
plt.title('Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. Region-wise Sales Analysis

```python
region_sales = sales_data.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
region_quantity = sales_data.groupby('Region')['Quantity'].sum().sort_values(ascending=False)
display(region_sales.to_frame('Total Sales'))
display(region_quantity.to_frame('Quantity Sold'))
print('Top Region:', region_sales.idxmax())
print('Top Region Sales:', region_sales.max())

plt.figure(figsize=(9,6))
region_sales.plot(kind='bar')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Sales Quantity and Daily Trend

```python
plt.figure(figsize=(9,6))
product_quantity.plot(kind='bar')
plt.title('Quantity Sold by Product')
plt.xlabel('Product')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

plt.figure(figsize=(9,6))
region_quantity.plot(kind='bar')
plt.title('Quantity Sold by Region')
plt.xlabel('Region')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

daily_sales = sales_data.groupby('Date')['Total_Sales'].sum()
highest_sales_date = daily_sales.idxmax()
highest_daily_sales = daily_sales.max()
print('Highest Sales Date:', highest_sales_date.strftime('%Y-%m-%d'))
print('Highest Daily Sales:', highest_daily_sales)

plt.figure(figsize=(12,6))
daily_sales.plot(kind='line')
plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Customer Churn Data Overview

```python
print('Customer Data - First 5 Rows')
display(customer_data.head())
print('\nMissing values:')
display(customer_data.isnull().sum())
print('\nDuplicate rows:', customer_data.duplicated().sum())
```

## 7. Overall Customer Churn

```python
total_customers = len(customer_data)
churned_customers = (customer_data['Churn'] == 1).sum()
non_churned_customers = (customer_data['Churn'] == 0).sum()
churn_rate = churned_customers / total_customers * 100

print('Total Customers:', total_customers)
print('Churned Customers:', churned_customers)
print('Non-Churned Customers:', non_churned_customers)
print('Churn Rate:', round(churn_rate, 2), '%')

churn_counts = customer_data['Churn'].value_counts().sort_index()
plt.figure(figsize=(8,6))
churn_counts.plot(kind='bar')
plt.title('Customer Churn Distribution')
plt.xlabel('Churn Status (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Churn by Contract and Payment Method

```python
contract_rates = customer_data.groupby('Contract')['Churn'].mean() * 100
payment_rates = customer_data.groupby('PaymentMethod')['Churn'].mean() * 100

print('Churn Rate by Contract (%)')
display(contract_rates.to_frame('Churn Rate (%)'))
print('Churn Rate by Payment Method (%)')
display(payment_rates.to_frame('Churn Rate (%)'))

plt.figure(figsize=(9,6))
contract_rates.plot(kind='bar')
plt.title('Churn Rate by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10,6))
payment_rates.plot(kind='bar')
plt.title('Churn Rate by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

## 9. Churn by Billing, Tenure and Charges

```python
paperless_rates = customer_data.groupby('PaperlessBilling')['Churn'].mean() * 100
senior_rates = customer_data.groupby('SeniorCitizen')['Churn'].mean() * 100
tenure_churn = customer_data.groupby('Churn')['Tenure'].mean()
monthly_charges_churn = customer_data.groupby('Churn')['MonthlyCharges'].mean()
total_charges_churn = customer_data.groupby('Churn')['TotalCharges'].mean()

print('Churn Rate by Paperless Billing (%)')
display(paperless_rates.to_frame('Churn Rate (%)'))
print('Churn Rate by Senior Citizen Status (%)')
display(senior_rates.to_frame('Churn Rate (%)'))
   
print('Average Tenure - Non-Churned:', round(tenure_churn[0], 2), 'months')
print('Average Tenure - Churned:', round(tenure_churn[1], 2), 'months')
print('Average Monthly Charges - Non-Churned:', round(monthly_charges_churn[0], 2))
print('Average Monthly Charges - Churned:', round(monthly_charges_churn[1], 2))
print('Average Total Charges - Non-Churned:', round(total_charges_churn[0], 2))
print('Average Total Charges - Churned:', round(total_charges_churn[1], 2))

tenure_groups = pd.cut(customer_data['Tenure'], bins=[0,12,24,36,48,60,72], labels=['0-12 months','13-24 months','25-36 months','37-48 months','49-60 months','61-72 months'])
tenure_churn_rate = customer_data.groupby(tenure_groups, observed=False)['Churn'].mean() * 100
display(tenure_churn_rate.to_frame('Churn Rate (%)'))

plt.figure(figsize=(10,6))
tenure_churn_rate.plot(kind='bar')
plt.title('Churn Rate by Tenure Group')
plt.xlabel('Tenure')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 10. Key Insights and Recommendations

**Key Insights**
- Laptops generated the highest sales among the products in the provided sales dataset.
- The North region generated the highest regional sales.
- The overall customer churn rate is 10.6% in the provided customer dataset.
- Month-to-month customers show the highest churn rate among the contract types.
- In this dataset, churn is concentrated among customers with 0–12 months of tenure.
- Churned customers have higher average monthly charges than non-churned customers.

**Recommendations**
- Promote high-performing products while improving performance in lower-sales regions.
- Encourage month-to-month customers to move to longer-term contracts using suitable retention offers.
- Strengthen onboarding and early-stage customer engagement during the first year.
- Review pricing and plan value for customers with higher monthly charges.
- Use churn monitoring and targeted retention campaigns to identify at-risk customers early.
